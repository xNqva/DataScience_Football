```python
def calculate_result(row):
    if row["FTHG"] > row["FTAG"]:
        return "H"
    elif row["FTHG"] < row["FTAG"]:
        return "A"
    else:
        return "D"
```

```python
calculated_results = df_clean.apply(
    calculate_result,
    axis=1
)
```

```python
(calculated_results != df_clean["FTR"]).sum()
```
