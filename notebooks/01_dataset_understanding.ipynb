```python
#Football variable
df["TotalGoals"] = df["FTHG"] + df["FTAG"]

df[
    [
        "HomeTeam",
        "AwayTeam",
        "FTHG",
        "FTAG",
        "TotalGoals"
    ]
].head(10)
```

```python
#First visualization
result_counts = df["FTR"].value_counts()

result_counts.index = result_counts.index.map({
    "H": "Home Win",
    "D": "Draw",
    "A": "Away Win"
})

plt.figure(figsize=(8, 5))

result_counts.plot(kind="bar")

plt.title("Premier League Match Results")
plt.xlabel("Result")
plt.ylabel("Number of Matches")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
#Goals per match
goal_distribution = df["TotalGoals"].value_counts().sort_index()
goal_distribution

plt.figure(figsize=(10, 5))

goal_distribution.plot(kind="bar")

plt.title("Number of Matches by Total Goals")
plt.xlabel("Total Goals")
plt.ylabel("Number of Matches")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
#Home vs Away Goals
home_goals = df["FTHG"].sum()
away_goals = df["FTAG"].sum()

print("Total home goals:", home_goals)
print("Total away goals:", away_goals)

goals = pd.Series({
    "Home": home_goals,
    "Away": away_goals
})

plt.figure(figsize=(7, 5))

goals.plot(kind="bar")

plt.title("Home Goals vs Away Goals")
plt.xlabel("Team Location")
plt.ylabel("Total Goals")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```
